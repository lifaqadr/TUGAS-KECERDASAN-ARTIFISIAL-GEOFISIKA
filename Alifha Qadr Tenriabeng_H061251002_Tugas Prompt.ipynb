# Prompt 1: Penentuan Hiposenter Gempa Bumi

## Role (R)
Kamu adalah seorang ahli seismologi yang berpengalaman dalam menentukan lokasi hiposenter gempa bumi menggunakan data waktu tiba gelombang seismik.

## Task (T)
Tentukan estimasi koordinat horizontal dan kedalaman hiposenter berdasarkan data waktu tiba gelombang P dan S dari beberapa stasiun seismik.

## Context (C)
Penentuan hiposenter diperlukan untuk mengetahui lokasi sumber gempa di bawah permukaan bumi. Analisis ini digunakan sebagai latihan mahasiswa Geofisika dalam memahami prinsip penentuan lokasi gempa.

## Input (I)
Gunakan data simulasi berikut:
- Jumlah stasiun: 4 stasiun.
- Kecepatan gelombang P: 6,2 km/s.
- Kecepatan gelombang S: 3,6 km/s.
- Waktu tiba gelombang P dan S dari setiap stasiun.
- Koordinat setiap stasiun.

Jika data waktu tiba atau koordinat stasiun belum tersedia, minta data tersebut sebelum melakukan perhitungan.

## Constraints (K)
- Jangan mengubah data waktu tiba yang diberikan.
- Gunakan satuan detik dan kilometer secara konsisten.
- Jelaskan asumsi model kecepatan yang digunakan.
- Bedakan estimasi awal dengan hasil inversi hiposenter.
- Jangan mengklaim hasil sebagai lokasi pasti jika data belum mencukupi.

## Output (O)
Sajikan:
1. Tabel data setiap stasiun.
2. Perhitungan selisih waktu tiba gelombang P dan S.
3. Penjelasan metode penentuan hiposenter.
4. Hasil estimasi lokasi dan kedalaman jika data mencukupi.
5. Visualisasi posisi stasiun dan hiposenter.
6. Kesimpulan.

## Quality Criteria (Q)
- Rumus dan perhitungan benar.
- Satuan konsisten.
- Asumsi dijelaskan.
- Hasil dapat ditelusuri dari data input.
- Visualisasi memiliki label dan satuan yang jelas.

## Optional Verification (V)
Periksa kembali waktu tiba, koordinat stasiun, dan hasil perhitungan. Jika memungkinkan, bandingkan hasil estimasi dengan metode numerik lain atau data referensi.

# Prompt 2: Analisis Gradien Geotermal

## Role (R)
Kamu adalah seorang geofisikawan yang berfokus pada studi panas bumi dan perpindahan panas di dalam bumi.

## Task (T)
Hitung gradien geotermal dari data suhu terhadap kedalaman dan jelaskan pola perubahan suhu bawah permukaan.

## Context (C)
Gradien geotermal menunjukkan perubahan suhu terhadap kedalaman. Informasi ini dapat membantu memahami kondisi termal bawah permukaan dan menjadi salah satu data pendukung studi panas bumi.

## Input (I)
Gunakan data simulasi berikut:

| Kedalaman (m) | Suhu (°C) |
|---:|---:|
| 100 | 28 |
| 300 | 35 |
| 500 | 43 |
| 700 | 52 |
| 900 | 60 |

## Constraints (K)
- Gunakan data yang tersedia tanpa mengubah nilainya.
- Gunakan satuan °C/km untuk gradien geotermal.
- Jelaskan rumus yang digunakan.
- Jangan menyimpulkan keberadaan reservoir panas bumi hanya berdasarkan gradien suhu.
- Bedakan hasil perhitungan dengan interpretasi geologi.

## Output (O)
Buat:
1. Tabel data suhu dan kedalaman.
2. Rumus gradien geotermal.
3. Perhitungan gradien pada setiap interval.
4. Nilai gradien rata-rata.
5. Grafik suhu terhadap kedalaman.
6. Interpretasi pola suhu.
7. Kesimpulan dan keterbatasan analisis.

## Quality Criteria (Q)
- Perhitungan memiliki satuan yang benar.
- Grafik menggunakan sumbu dan label yang sesuai.
- Interpretasi didukung data.
- Hasil perhitungan dapat diperiksa ulang.
- Keterbatasan data dijelaskan.

## Optional Verification (V)
Hitung kembali gradien rata-rata menggunakan perubahan suhu total dibagi perubahan kedalaman total. Bandingkan hasilnya dengan perhitungan per interval.

# Prompt 3: Analisis Frekuensi Sinyal Seismik

## Role (R)
Kamu adalah seorang ahli pemrosesan sinyal geofisika yang memahami analisis sinyal seismik menggunakan transformasi Fourier.

## Task (T)
Analisis sinyal seismik dalam domain waktu dan ubah sinyal tersebut ke domain frekuensi menggunakan Fast Fourier Transform (FFT).

## Context (C)
Sinyal seismik dapat dianalisis dalam domain waktu maupun frekuensi. Analisis frekuensi membantu mengidentifikasi komponen frekuensi dominan dan karakteristik sinyal.

## Input (I)
Gunakan parameter simulasi berikut:
- Frekuensi sampling: 100 Hz.
- Durasi sinyal: 10 detik.
- Komponen sinyal: 5 Hz dan 15 Hz.
- Amplitudo komponen pertama: 2.
- Amplitudo komponen kedua: 0,8.
- Tambahkan noise acak dengan tingkat yang dapat diatur.

Jika data sinyal asli tersedia, gunakan data tersebut sebagai pengganti sinyal simulasi.

## Constraints (K)
- Gunakan Python dengan NumPy dan Matplotlib.
- Jelaskan fungsi setiap bagian kode.
- Jangan menyatakan frekuensi dominan tanpa memeriksa spektrum.
- Bedakan sinyal simulasi dengan sinyal pengukuran asli.
- Perhatikan frekuensi Nyquist dan normalisasi amplitudo spektrum.

## Output (O)
Sajikan:
1. Penjelasan teori FFT secara sederhana.
2. Kode Python yang dapat dijalankan.
3. Grafik sinyal terhadap waktu.
4. Grafik spektrum amplitudo terhadap frekuensi.
5. Identifikasi frekuensi dominan.
6. Interpretasi hasil.
7. Kesimpulan.

## Quality Criteria (Q)
- Kode dapat dijalankan.
- Sumbu waktu dan frekuensi diberi label.
- Satuan dan frekuensi sampling digunakan dengan benar.
- Grafik mudah dibaca.
- Hasil interpretasi sesuai spektrum.

## Optional Verification (V)
Uji program dengan sinyal yang memiliki frekuensi diketahui, yaitu 5 Hz dan 15 Hz. Pastikan puncak spektrum berada di sekitar frekuensi tersebut dengan mempertimbangkan resolusi frekuensi.

# Prompt 4: Interpretasi Data Resistivitas Bawah Permukaan

## Role (R)
Kamu adalah seorang geofisikawan yang ahli dalam metode geolistrik resistivitas dan interpretasi lapisan bawah permukaan.

## Task (T)
Analisis data resistivitas untuk mengidentifikasi perubahan sifat kelistrikan bawah permukaan dan menyusun interpretasi awal lapisan tanah atau batuan.

## Context (C)
Metode geolistrik memanfaatkan perbedaan sifat kelistrikan material untuk memperoleh informasi tentang kondisi bawah permukaan. Nilai resistivitas dapat dipengaruhi oleh jenis batuan, porositas, kandungan fluida, salinitas, dan kondisi geologi lainnya.

## Input (I)
Gunakan data simulasi berikut:

| Lapisan | Ketebalan (m) | Resistivitas (Ω·m) |
|---|---:|---:|
| 1 | 2,5 | 120 |
| 2 | 6,0 | 35 |
| 3 | 11,0 | 240 |
| 4 | Tidak diketahui | 650 |

## Constraints (K)
- Pertahankan nilai data simulasi.
- Jangan menyamakan satu nilai resistivitas dengan satu jenis batuan secara pasti.
- Jelaskan bahwa interpretasi memerlukan informasi geologi pendukung.
- Jangan menganggap tabel ini sebagai hasil inversi lapangan.
- Nyatakan bahwa kedalaman lapisan terakhir belum diketahui.

## Output (O)
Buat:
1. Tabel data resistivitas.
2. Grafik model resistivitas terhadap kedalaman jika dapat disusun dari data.
3. Deskripsi perubahan resistivitas antar lapisan.
4. Interpretasi kemungkinan kondisi material bawah permukaan.
5. Penjelasan ketidakpastian.
6. Kesimpulan.

## Quality Criteria (Q)
- Satuan resistivitas benar.
- Kedalaman dihitung secara konsisten.
- Data yang tidak diketahui tidak dibuat-buat.
- Interpretasi tidak berlebihan.
- Grafik dan tabel sesuai dengan data.

## Optional Verification (V)
Periksa kembali penjumlahan ketebalan lapisan untuk menentukan kedalaman batas lapisan. Pastikan lapisan dengan ketebalan yang tidak diketahui tidak diberi batas kedalaman fiktif.

# Prompt 5: Analisis Anomali Gravitasi

## Role (R)
Kamu adalah seorang geofisikawan yang berpengalaman dalam metode gravitasi untuk mempelajari variasi densitas material bawah permukaan.

## Task (T)
Analisis data gravitasi untuk mengidentifikasi pola anomali dan menjelaskan kemungkinan penyebab geologi berdasarkan data yang tersedia.

## Context (C)
Metode gravitasi mempelajari variasi medan gravitasi yang dapat berkaitan dengan perbedaan distribusi massa atau densitas bawah permukaan. Interpretasi perlu mempertimbangkan koreksi pengukuran dan kondisi geologi.

## Input (I)
Gunakan data simulasi berikut:

| Titik | Koordinat X (m) | Anomali Bouguer (mGal) |
|---|---:|---:|
| A | 0 | -12,4 |
| B | 250 | -8,7 |
| C | 500 | -3,2 |
| D | 750 | 2,1 |
| E | 1000 | 5,8 |

## Constraints (K)
- Gunakan nilai data tanpa perubahan.
- Jangan menghitung ulang koreksi Bouguer jika data koreksi yang dibutuhkan tidak tersedia.
- Bedakan antara pola anomali dan interpretasi penyebabnya.
- Jangan menentukan jenis batuan atau struktur secara pasti hanya dari data ini.
- Jelaskan keterbatasan interpretasi satu lintasan pengukuran.

## Output (O)
Sajikan:
1. Tabel data awal.
2. Grafik anomali Bouguer terhadap koordinat X.
3. Deskripsi pola anomali.
4. Kemungkinan interpretasi geologi.
5. Keterbatasan data.
6. Kesimpulan.

## Quality Criteria (Q)
- Grafik memiliki judul, label sumbu, dan satuan.
- Data tidak berubah.
- Interpretasi sesuai pola anomali.
- Hasil pengamatan dipisahkan dari dugaan geologi.
- Kesimpulan tidak melebihi informasi yang tersedia.

## Optional Verification (V)
Periksa kembali seluruh nilai pada grafik terhadap tabel. Pastikan tanda negatif dan positif tidak tertukar dan koordinat X diurutkan dengan benar.
